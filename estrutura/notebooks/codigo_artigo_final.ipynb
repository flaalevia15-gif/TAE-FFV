```r

# Anos que serão utilizados para validação
anos_validacao <- c(2022, 2023, 2024)

# Criar tabela para armazenar os resultados
resultados_cv <- data.frame()

# Armazenar previsões de cada rodada
previsoes_cv <- data.frame()

# Repetir o treinamento para cada ano
for (ano in anos_validacao) {

  # Separar treinamento e validação
  treino_fold <- subset(
    treino_rf,
    Ano < ano
  )

  validacao_fold <- subset(
    treino_rf,
    Ano == ano
  )

  # Verificar se existem categorias novas
  categorias_novas <- c(
    setdiff(
      unique(as.character(validacao_fold$Municipio)),
      unique(as.character(treino_fold$Municipio))
    ),
    setdiff(
      unique(as.character(validacao_fold$Codigo_SH4)),
      unique(as.character(treino_fold$Codigo_SH4))
    ),
    setdiff(
      unique(as.character(validacao_fold$mes_numero)),
      unique(as.character(treino_fold$mes_numero))
    )
  )

  if (length(categorias_novas) > 0) {
    stop(paste(
      "Categoria não encontrada no treinamento do ano",
      ano
    ))
  }

  # Treinar Random Forest
  set.seed(123)

  modelo_fold <- randomForest(
    Quilograma_Liquido ~
      Municipio +
      mes_numero +
      Codigo_SH4 +
      Ano,
    data = treino_fold,
    ntree = 500
  )

  # Fazer previsões
  prev_fold <- predict(
    modelo_fold,
    newdata = validacao_fold
  )

  # Calcular métricas
  rmse_fold <- sqrt(
    mean(
      (validacao_fold$Quilograma_Liquido -
        prev_fold)^2
    )
  )

  mae_fold <- mean(
    abs(
      validacao_fold$Quilograma_Liquido -
        prev_fold
    )
  )

  r2_fold <- 1 -
    sum(
      (validacao_fold$Quilograma_Liquido -
        prev_fold)^2
    ) /
    sum(
      (validacao_fold$Quilograma_Liquido -
        mean(validacao_fold$Quilograma_Liquido))^2
    )

  # Guardar resultados
  resultados_cv <- rbind(
    resultados_cv,
    data.frame(
      Ano = ano,
      RMSE = rmse_fold,
      MAE = mae_fold,
      R2 = r2_fold
    )
  )

  # Guardar previsões
  previsoes_cv <- rbind(
    previsoes_cv,
    data.frame(
      Ano = ano,
      Real = validacao_fold$Quilograma_Liquido,
      Previsto = prev_fold
    )
  )
}

# Mostrar resultados de cada rodada
resultados_cv
```

VOLTAR DAQUI AMANHÃ

```r

# Quantidade de reamostragens
B <- 1000

# Organizar os meses em ordem cronológica
meses <- sort(unique(
  format(previsoes_boot$Data, "%Y-%m")
))

# Quantidade de meses
n_meses <- length(meses)

# Função para calcular as métricas
calcular_metricas <- function(amostra) {

  real <- amostra$Real

  prev_lm <- amostra$Previsto_LM
  prev_rf <- amostra$Previsto_RF

  # Métricas da regressão linear
  rmse_lm <- sqrt(mean((real - prev_lm)^2))
  mae_lm <- mean(abs(real - prev_lm))

  r2_lm <- 1 -
    sum((real - prev_lm)^2) /
    sum((real - mean(real))^2)

  # Métricas do Random Forest
  rmse_rf <- sqrt(mean((real - prev_rf)^2))
  mae_rf <- mean(abs(real - prev_rf))

  r2_rf <- 1 -
    sum((real - prev_rf)^2) /
    sum((real - mean(real))^2)

  # Diferença entre os modelos
  c(
    RMSE_LM = rmse_lm,
    MAE_LM = mae_lm,
    R2_LM = r2_lm,
    RMSE_RF = rmse_rf,
    MAE_RF = mae_rf,
    R2_RF = r2_rf,
    Dif_RMSE = rmse_rf - rmse_lm
  )
}

# Função para realizar Bootstrap em blocos
bootstrap_blocos <- function(tamanho_bloco, B = 1000) {

  resultados <- matrix(
    NA,
    nrow = B,
    ncol = 7
  )

  colnames(resultados) <- c(
    "RMSE_LM", "MAE_LM", "R2_LM",
    "RMSE_RF", "MAE_RF", "R2_RF",
    "Dif_RMSE"
  )

  set.seed(123)

  for (b in 1:B) {

    meses_amostra <- character(0)

    # Sortear blocos consecutivos
    while (length(meses_amostra) < n_meses) {

      inicio <- sample(
        1:(n_meses - tamanho_bloco + 1),
        1
      )

      bloco <- meses[
        inicio:(inicio + tamanho_bloco - 1)
      ]

      meses_amostra <- c(
        meses_amostra,
        bloco
      )
    }

    # Manter exatamente 36 meses
    meses_amostra <- meses_amostra[1:n_meses]

    # Recuperar todas as observações dos meses sorteados
    indices <- unlist(
      lapply(meses_amostra, function(m) {
        which(
          format(previsoes_boot$Data, "%Y-%m") == m
        )
      })
    )

    amostra <- previsoes_boot[indices, ]

    # Calcular métricas
    resultados[b, ] <- calcular_metricas(amostra)
  }

  return(resultados)
}

# Bootstrap com blocos de 3 meses
boot_3 <- bootstrap_blocos(3, B)

# Bootstrap com blocos de 6 meses
boot_6 <- bootstrap_blocos(6, B)

# Conferir resultados
head(boot_3)
head(boot_6)
```

```r

# Intervalos de confiança - blocos de 3 meses
ic_3 <- apply(
  boot_3,
  2,
  quantile,
  probs = c(0.025, 0.975),
  na.rm = TRUE
)

# Intervalos de confiança - blocos de 6 meses
ic_6 <- apply(
  boot_6,
  2,
  quantile,
  probs = c(0.025, 0.975),
  na.rm = TRUE
)

# Mostrar resultados
ic_3
ic_6
```

```r

# Validacao cruzada temporal dos modelos separados

library(randomForest)

set.seed(123)

anos_validacao <- c(2022, 2023, 2024)

resultados_cv_separados <- data.frame()

previsoes_cv_separados <- data.frame()

for (fluxo_atual in c("Exportação", "Importação")) {

  dados_fluxo <- treino_rf[
    treino_rf$Fluxo == fluxo_atual, ]

  for (ano_atual in anos_validacao) {

    # Separar treino e validacao
    dados_treino <- dados_fluxo[
      dados_fluxo$Ano < ano_atual, ]

    dados_validacao <- dados_fluxo[
      dados_fluxo$Ano == ano_atual, ]

    dados_treino <- droplevels(dados_treino)

    # Preditoras
    variaveis <- c(
      "Municipio",
      "mes_numero",
      "Codigo_SH4",
      "Ano"
    )

    # Remover fatores com apenas um nivel
    variaveis <- variaveis[
      sapply(variaveis, function(v) {
        if (is.factor(dados_treino[[v]])) {
          nlevels(dados_treino[[v]]) >= 2
        } else {
          TRUE
        }
      })
    ]

    # Ajustar niveis dos fatores
    for (v in variaveis) {
      if (is.factor(dados_treino[[v]])) {

        dados_validacao[[v]] <- factor(
          as.character(dados_validacao[[v]]),
          levels = levels(dados_treino[[v]])
        )
      }
    }

    # Verificar categorias desconhecidas
    if (anyNA(dados_validacao[, variaveis])) {
      stop(paste(
        "Categoria desconhecida:",
        fluxo_atual, ano_atual
      ))
    }

    # Construir formula
    formula_rf <- as.formula(
      paste(
        "Quilograma_Liquido ~",
        paste(variaveis, collapse = " + ")
      )
    )

    # Treinar Random Forest
    modelo_fold <- randomForest(
      formula_rf,
      data = dados_treino,
      ntree = 500
    )

    # Prever
    previsto <- predict(
      modelo_fold,
      newdata = dados_validacao
    )

    real <- dados_validacao$Quilograma_Liquido

    # Metricas
    rmse <- sqrt(mean((real - previsto)^2))
    mae <- mean(abs(real - previsto))

    r2 <- 1 - sum((real - previsto)^2) /
      sum((real - mean(real))^2)

    # Armazenar resultados
    resultados_cv_separados <- rbind(
      resultados_cv_separados,
      data.frame(
        Ano = ano_atual,
        Fluxo = fluxo_atual,
        RMSE = rmse,
        MAE = mae,
        R2 = r2
      )
    )

    previsoes_cv_separados <- rbind(
      previsoes_cv_separados,
      data.frame(
        Ano = ano_atual,
        Fluxo = fluxo_atual,
        Real = real,
        Previsto = previsto
      )
    )
  }
}

# Resultados por ano e fluxo
resultados_cv_separados

# Previsoes
head(previsoes_cv_separados)
```

```r

# Consolidar resultados dos tres folds por fluxo

resumo_cv <- do.call(
  rbind,
  lapply(
    split(previsoes_cv_separados,
          previsoes_cv_separados$Fluxo),
    function(dados) {

      data.frame(
        Fluxo = as.character(dados$Fluxo[1]),
        Observacoes = nrow(dados),

        RMSE = sqrt(mean(
          (dados$Real - dados$Previsto)^2
        )),

        MAE = mean(abs(
          dados$Real - dados$Previsto
        )),

        R2 = 1 - sum(
          (dados$Real - dados$Previsto)^2
        ) / sum(
          (dados$Real - mean(dados$Real))^2
        )
      )
    }
  )
)

resumo_cv
```

```r

# Bootstrap das metricas dos modelos separados

set.seed(123)

B <- 1000

bootstrap_resultados <- data.frame()

for (fluxo_atual in c("Exportação", "Importação")) {

  dados <- previsoes_cv_separados[
    previsoes_cv_separados$Fluxo == fluxo_atual, ]

  resultados_boot <- replicate(B, {

    indices <- sample(
      1:nrow(dados),
      replace = TRUE
    )

    amostra <- dados[indices, ]

    real <- amostra$Real
    previsto <- amostra$Previsto

    rmse <- sqrt(mean((real - previsto)^2))

    mae <- mean(abs(real - previsto))

    r2 <- 1 - sum((real - previsto)^2) /
      sum((real - mean(real))^2)

    c(RMSE = rmse, MAE = mae, R2 = r2)
  })

  intervalo <- t(apply(
    resultados_boot,
    1,
    quantile,
    probs = c(0.025, 0.975),
    na.rm = TRUE
  ))

  bootstrap_resultados <- rbind(
    bootstrap_resultados,
    data.frame(
      Fluxo = fluxo_atual,
      Indicador = rownames(intervalo),
      Limite_Inferior = intervalo[, 1],
      Limite_Superior = intervalo[, 2]
    )
  )
}

bootstrap_resultados
```

```r

# Bootstrap temporal em blocos de 3 meses

set.seed(123)

B <- 1000
tamanho_bloco <- 3

resultado_boot_temporal <- data.frame()

for (fluxo_atual in c("Exportação", "Importação")) {

  dados <- previsoes_cv_separados[
    previsoes_cv_separados$Fluxo == fluxo_atual, ]

  meses <- sort(unique(dados$Data))

  n_meses <- length(meses)

  n_blocos <- ceiling(n_meses / tamanho_bloco)

  resultados <- replicate(B, {

    inicios <- sample(
      1:(n_meses - tamanho_bloco + 1),
      n_blocos,
      replace = TRUE
    )

    amostra <- do.call(
      rbind,
      lapply(inicios, function(i) {

        meses_bloco <- meses[
          i:(i + tamanho_bloco - 1)
        ]

        dados[dados$Data %in% meses_bloco, ]
      })
    )

    real <- amostra$Real
    previsto <- amostra$Previsto

    rmse <- sqrt(mean((real - previsto)^2))

    mae <- mean(abs(real - previsto))

    r2 <- 1 - sum((real - previsto)^2) /
      sum((real - mean(real))^2)

    c(RMSE = rmse, MAE = mae, R2 = r2)
  })

  intervalos <- t(apply(
    resultados,
    1,
    quantile,
    probs = c(0.025, 0.975),
    na.rm = TRUE
  ))

  resultado_boot_temporal <- rbind(
    resultado_boot_temporal,

    data.frame(
      Fluxo = fluxo_atual,
      Indicador = rownames(intervalos),
      Limite_Inferior = intervalos[, 1],
      Limite_Superior = intervalos[, 2]
    )
  )
}

resultado_boot_temporal
```
